```python
#OOP Programming
class Solution:
    def sort012(self, arr):
        zeros = []
        ones = []
        twos = []

        for num in arr:
            if num == 0:
                zeros.append(num)
            if num == 1:
                ones.append(num)
            if num == 2:
                twos.append(num)

        arr[:] = zeros+ones+twos
        print(arr)

solu = Solution()
solu.sort012([0,0,0,2,1,2,1,0,2,2,0,1,1,2,2,])
```
