```python
#OOP Programming
class Solution:
    def sort012(self, arr):
        zeros = []
        ones = []
        twos = []

        for num in arr:
            if num == 0:
                zeros.append(num)
            if num == 1:
                ones.append(num)
            if num == 2:
                twos.append(num)

        arr[:] = zeros+ones+twos
        print(arr)

solu = Solution()
solu.sort012([0,0,0,2,1,2,1,0,2,2,0,1,1,2,2,])
```

```python
#Functional Programming
class Solution:
    def firstSearch(self, arr, k):
        left = 0
        right = len(arr) - 1
        idx = -1
        while left <= right:
            middle = (left + right) // 2
            
            if arr[middle] == k:   
                idx = middle
                right = middle - 1
            elif arr[middle] < k:
                left = middle + 1
            else:
                right = middle - 1
        return idx
                
solu = Solution()
solu.firstSearch([1,3,6,8,20], 2)        
```
